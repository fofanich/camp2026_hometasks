# floors & glasses

```python
def max_floors(drops, glasses):
    if drops == 0: return 0
    if glasses == 1: return drops

    # if glass breaks, check floors below
    floors_if_glass_broken = max_floors(drops - 1, glasses - 1)

    # if glass survives, check floors above
    floors_if_glass_survived = max_floors(drops - 1, glasses)

    # return total floors covered: below + current + above
    return floors_if_glass_broken + 1 + floors_if_glass_survived

def min_attempts(floors, glasses):
    drops = 0
    # increase drops until all floors with given glasses can be checked
    while max_floors(drops, glasses) < floors:
        drops += 1
    return drops

# examples for testing
test_cases = [(10, 2),(100, 2),(100, 3),(1000, 3),(1000, 5)]

for floors, glasses in test_cases:
    result = min_attempts(floors, glasses)
    print(f"{floors} floors, {glasses} glasses => {result} attempts")
```
