# Solution for sum and product problem 
1) P sees only the product
if multiple pairs share the same product — P can't tell which one it is
p=12 => (2,6) and (3,4), so P doesn't know
2) S sees only the sum, yet claims he already knew P didn't know
that's only possible if every pair with this sum has a non-unique product
if even one pair gives a unique product — P could have known, so S couldn't be sure
sum=7 => (2,5) gives product 10 which is unique -> S couldn't be certain -> sum 7 fails here
3) after S spoke, P gets new info: S sum is "safe"
P filters his candidates — keeps only pairs whose sum passes step 2
if exactly one survives — P now knows
4) after P spoke, S knows P narrowed it down to one
S filters his candidates — keeps only pairs that pass step 2 and step 3
if exactly one survives — S knows too



```python
# with rules that given i create all possible pairs (x,y), 
candidates = [
    (x, y)
    for x in range(2, 100)
    for y in range(x + 1, 100)
    if x + y <= 100
]

# helpful lookups to avoid scanning all candidates every time
# by_product[12] = [(2,6), (3,4)]
by_product = {}
for pair in candidates:
    p = pair[0] * pair[1]
    if p not in by_product:
        by_product[p] = []
    by_product[p].append(pair)

# by_sum[7] = [(2,5), (3,4)]
by_sum = {}
for pair in candidates:
    s = pair[0] + pair[1]
    if s not in by_sum:
        by_sum[s] = []
    by_sum[s].append(pair)

def p_not_know(pair):
    # P checks if this product has multiple possible pairs; 
    # if yes, P cannot know the numbers.
    p = pair[0] * pair[1]
    return len(by_product[p]) > 1

def s_knew_p_not_know(pair):
    # S checks all pairs with the same sum; 
    # if any gives a unique product, 
    # S could not be sure, so step 2 fails
    s = pair[0] + pair[1]
    for c in by_sum[s]: 
        if not p_not_know(c):
            return False
    return True
    
def p_now_knows(pair):
    # After S statement, 
    # P filters product-candidates by step 2; 
    # if one candidate remains, P now knows
    p = pair[0] * pair[1]
    surviving = [c for c in by_product[p] if s_knew_p_not_know(c)]
    return len(surviving) == 1

def s_now_knows(pair):
    # After P’s statement, 
    # S filters sum-candidates by step 2 and step 3; 
    # if one candidate remains, S now knows.
    s = pair[0] + pair[1]
    surviving = [c for c in by_sum[s] if s_knew_p_not_know(c) and p_now_knows(c)]
    return len(surviving) == 1

answer = [
    c for c in candidates
    if s_knew_p_not_know(c)
    and p_now_knows(c)
    and s_now_knows(c)
]

print(f"answer: {answer}")
```
